```python
import pandas as pd

df = pd.read_csv(
    'Data_Sumatera_Barat_Kotor.csv',
    header=1
)

df.columns = df.columns.str.strip()

df['Unnamed: 0'] = df['Unnamed: 0'].astype(str).str.strip()

data_sumbar = df[
    df['Unnamed: 0'] == 'Provinsi Sumatera Barat'
]


laki_laki = float(data_sumbar['Laki-Laki'].values[0]) * 1000
perempuan = float(data_sumbar['Perempuan'].values[0]) * 1000


total_keseluruhan = laki_laki + perempuan


hasil = pd.DataFrame({
    'provinsi': ['Sumatera Barat'],
    'laki_laki': [int(laki_laki)],
    'perempuan': [int(perempuan)],
    'total_keseluruhan': [int(total_keseluruhan)]
})


print(hasil)


hasil.to_csv(
    'jumlah_penduduk_sumatera_barat.csv',
    index=False
)


hasil.to_json(
    'jumlah_penduduk_sumatera_barat.json',
    orient='records',
    indent=4
)

print("CSV dan JSON berhasil disimpan!")
```
